```python
from selenium import webdriver
from selenium.webdriver.common.by import By
driver = webdriver.Chrome()
scraped_data = []
for i in range(1, 51):
    driver.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
    books = driver.find_elements(By.CLASS_NAME, "product_pod")
    for book in books:
        title = book.get_attribute("title")
        price = book.find_element(By.CLASS_NAME, "product_price").text
        availability = book.find_element(By.CLASS_NAME, "icon-ok").text
        #rating = book.find_element(By.CLASS_NAME, "")
        scraped_data.append({
            "title": title,
            "price": price,
            "availability": availability
        })
driver.quit()
```
